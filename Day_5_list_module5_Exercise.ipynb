## Task 1

**Declaring variable ages**

    . Sort the list and find the min and max age
    . Add the min age and the max age again to the list
    . Find the median age (one middle item or two middle items divided by two)
    . Find the average age (sum of all items divided by their number )
    . Find the range of the ages (max minus min)
    . Compare the value of (min - average) and (max - average), use abs() method

```python
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]
```

```python
# Sorting list and finding min & max age
print(ages)
ages.sort()
print(ages)

print(f'The minimum age is: {min(ages)}, and the maximum age is: {max(ages)}')
```

```python
# Adding min and max age to the list

ages.append(19)
ages.append(26)

print(ages)
```

```python
# Find the median age (one middle item or two middle items divided by two)

print(len(ages))
age_med = (24 + 24) // 2
print("The median is:", age_med)
```

```python
# Find the average age (sum of all items divided by their number )

avg_ages = sum(ages) / len(ages)
print(avg_ages)
```

```python
# Find the range of the ages (max minus min)

age_range = max(ages) - min(ages)
print(age_range)
```

```python
# Compare the value of (min - average) and (max - average), use abs() method

min_avg = abs(min(ages) - avg_ages)
max_avg = abs(max(ages) - avg_ages)

print(min_avg)
print(max_avg)
```

## Task 3

**I am a teacher and I love to inspire and teach people. How many unique words have been used in the sentence? Use the split methods and set to get the unique words.**

```python
sentense = "I am a teacher and I love to inspire and teach people"
words = sentense.split()
sentense_2set = set(words)
print(sentense_2set)
print(f"The sentense contains total of {len(sentense_2set)} unique words")
print(sentense_2set)
```
