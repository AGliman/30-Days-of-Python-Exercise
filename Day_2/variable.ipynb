```python
#Calculating Addition, Subtraction, Multiplication, Division, Modulus, Exponential and Floor division of two numbers

num_one = 5
num_two = 4
sum = num_one + num_two
total = sum
print(f'The sum of {num_one} and {num_two} is {total}')

sub = num_one - num_two
diff = sub
print(f'The difference of {num_one} and {num_two} is {diff}')

mul = num_one * num_two
product = mul
print(f'The product of {num_one} and {num_two} is {product}')

div = num_one / num_two
division = div
print(f'The division of {num_one} and {num_two} is {division}')

modu =  num_one % num_two
reminder = modu
print(f'The remainder of {num_one} divided by {num_two} is {reminder}')

pow = num_one ** num_two
exp = pow
print(f'The result of {num_one} raised to the power of {num_two} is {exp}')

floor = num_one // num_two
floor_division = floor
print(f'The floor division of {num_one} by {num_two} is {floor_division}')
```

```python
#Calculating area of a circle
radius = 30

pi = 3.24
area_of_circle = pi * radius ** 2
print(f'The area of a circle with a radius of {radius} is {area_of_circle}')
```

```python
#Calculating Circumference of a Circle
circum_of_circle = 2 * pi * radius
print("The circumference of a circle with a radius", radius, "is: ", circum_of_circle)
```

```python
#Take radius as user input and calculate the area
radius = float(input("Enter the radius of a circle: "))
area = pi * radius ** 2
print("The area of a Circle with a radius", radius, "is: ", area)
```
