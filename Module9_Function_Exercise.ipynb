**Write a function called check-season, it takes a month parameter and returns the season: Autumn, Winter, Spring or Summer.**

```python
def check_season(month):
    if month in ["December", "January", "February"]:
        return "Winter"
    elif month in ["March", "April", "May"]:
        return "Spring"
    elif month in ["June", "July", "August"]:
        return "Summer"
    elif month in ["September", "October", "November"]:
        return "Autumn"
    else:
        return "Invalid month"
    
print("We are in",check_season("May"), "season")
```

### Task 7

**Quadratic equation is calculated as follows: ax² + bx + c = 0. Write a function which calculates solution set of a quadratic equation, solve_quadratic_eqn.**


```python
import math
def quadratic_equation(a, b, c):
    d = (b ** 2) - 4 * a * c

    sqrt_val = math.sqrt(abs(d))

    x1 = (-b + sqrt_val) / (2 * a)
    x2 = (-b - sqrt_val) / (2 * a)

    return x1, x2
quadratic_equation(1, -3, 2)
```

**Declare a function named remove_item. It takes a list and an item parameters. It returns a list with the item removed from it.**

    numbers = [2, 3, 7, 9]
    print(remove_item(numbers, 3))  # [2, 7, 9

```python
def remove_item(lst, item):
    new_list = []
    removed = False
    for x in lst:
        if x == item and not removed:
            removed = True
            continue
        new_list.append(x)
    return new_list

# Example
food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(remove_item(food_staff, 'Mango'))


numbers = [2, 3, 7, 9]
print(remove_item(numbers, 3))
```

**Declare a function named sum_of_odds. It takes a number parameter and it adds all the odd numbers in that range.**

```python
def sum_of_odds(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i
    return total

print(sum_of_odds(5))   
print(sum_of_odds(10))  
```

    print(evens_and_odds(100))
    # The number of odds are 50.
    # The number of evens are 51.

```python
def evens_and_odds(n):
    evens = 0
    odds = 0
    for i in range(n + 1):  # 0 to n inclusive
        if i % 2 == 0:
            evens += 1
        else:
            odds += 1
    return f"The number of odds are {odds}.\nThe number of evens are {evens}."

print(evens_and_odds(100))
```

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


print(factorial(5))
print(factorial(10))
```

### Task 4

**Write different functions which take lists. They should calculate_mean, calculate_median, calculate_mode, calculate_range, calculate_variance, calculate_std (standard deviation).**

```python
import math
from collections import Counter

def calculate_mean(lst):
    total = 0
    for x in lst:
        total += x
    return total / len(lst)

def calculate_median(lst):
    sorted_lst = sorted(lst)
    n = len(sorted_lst)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_lst[mid - 1] + sorted_lst[mid]) / 2
    else:
        return sorted_lst[mid]

def calculate_mode(lst):
    counts = Counter(lst)
    max_count = max(counts.values())
    modes = [k for k, v in counts.items() if v == max_count]
    return modes[0] if len(modes) == 1 else modes

def calculate_range(lst):
    return max(lst) - min(lst)

def calculate_variance(lst):
    mean = calculate_mean(lst)
    total = 0
    for x in lst:
        total += (x - mean) ** 2
    return total / len(lst)  # population variance

def calculate_std(lst):
    return math.sqrt(calculate_variance(lst))

# Example
data = [1, 2, 2, 3, 4, 5, 5, 5, 8, 9]

print(calculate_mean(data))     # 4.4
print(calculate_median(data))   # 4.5
print(calculate_mode(data))     # 5
print(calculate_range(data))    # 8
print(calculate_variance(data)) # 6.44
print(calculate_std(data))      # 2.537...
```

```python
#Write a function called is_prime, which checks if a number is prime.
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
is_prime(11)
```

```python
#Create a function called the most_populated_countries. It should return 10 or 20 most populated countries in descending order.
def most_populated_countries(countries, n=10):
    countries_left = countries.copy()
    top_countries = []

    while countries_left and len(top_countries) < n:
        most_populated = countries_left[0]

        for country in countries_left:
            if country["population"] > most_populated["population"]:
                most_populated = country

        top_countries.append(most_populated)
        countries_left.remove(most_populated)

    return top_countries

most_populated = most_populated_countries(countries_data, 10)
for country in most_populated:
    print(country["name"], country["population"])
```
