### Task 2

**Explain the difference between higher order function, closure and decorator**

**Answer**


*Higher Order Function:* A function that works with other functions - it either takes a function as input, or gives back a function as output.

*Closure:* A function that remembers the variables around it, even after that surrounding function has finished running.

*Decorator:* A function that wraps another function to add extra behavior to it, without changing the original function's code.



### Task 3

**Define a call function before map, filter or reduce, see examples.**

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

def double(number):
    return number * 2

print(list(map(double, numbers)))  


def is_even(number):
    return number % 2 == 0

print(list(filter(is_even, numbers)))  


def add(total, number):
    return total + number

print(reduce(add, numbers))  # 15
```
